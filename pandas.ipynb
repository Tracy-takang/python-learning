```python
def calculate_pay(name, hours):

    regular_rate = 3000
    overtime_rate = regular_rate * 1.5

    regular_pay = 0
    overtime_pay = 0
    total_pay = 0

    if hours <= 40:
        regular_pay = hours * regular_rate
    else:
        regular_pay = 40 * regular_rate
        overtime_pay = (hours - 40) * overtime_rate

    total_pay = regular_pay + overtime_pay

    print("Employee:", name)
    print("Regular Pay:", regular_pay)
    print("Overtime Pay:", overtime_pay)
    print("Total Pay:", total_pay)
    print("----------------------")


calculate_pay("Paul", 35)
calculate_pay("Mary", 40)
calculate_pay("Joseph", 46)
```
